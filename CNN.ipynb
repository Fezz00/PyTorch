```python
import numpy as np
from torchvision import datasets, transforms
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
import os
import matplotlib.pyplot as plt
```

```python
if(torch.cuda.is_available()):
    torch.device = "cuda"
elif(torch.backends.mps.is_available()):
    torch.device = "mps"
else:
    torch.device = "cpu"
```

```python
data_dir = os.path.expanduser("~/data")

train_data = datasets.MNIST(
    root=data_dir,
    train=True,
    download=True,
    transform=transforms.ToTensor(),
)
test_data = datasets.MNIST(
    root=data_dir,
    train=False,
    download=True,
    transform=transforms.ToTensor(),
)

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64, shuffle=True)

images, labels = next(iter(train_loader))
test_images, test_labels = next(iter(test_loader))

print(images.shape)
print(labels.shape)
```

```python
model = nn.Sequential(
    nn.Conv2d(1, 16, 3),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Conv2d(16, 32, 3),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Flatten(),
    nn.Linear(800, 10)
)

model.zero_grad()
z = model(images)
print(z.shape)
```

```python
model = nn.Sequential(
    nn.Conv2d(1, 32, 3),   # 28 - 3 + 1 = 26
    nn.BatchNorm2d(32),
    nn.ReLU(),
    nn.MaxPool2d(2),       # 26 / 2 = 13
    nn.Conv2d(32, 64, 3),  # 13 - 3 + 1 = 11
    nn.BatchNorm2d(64),
    nn.ReLU(),
    nn.MaxPool2d(2),       # 11 / 2 = 5 (floor)
    nn.Flatten(),
    nn.Linear(1600, 128),  # 64 * 5 * 5 = 1600
    nn.ReLU(),
    nn.Dropout(0.2),
    nn.Linear(128, 10)
)

model.to(device)

loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=0.001)

epochs = 5
total_steps = epochs * len(train_loader)
warmup_steps = 500

warmup = torch.optim.lr_scheduler.LinearLR(
    optimizer, start_factor=0.01, end_factor=1.0, total_iters=warmup_steps
)

cosine = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=total_steps - warmup_steps
)

scheduler = torch.optim.lr_scheduler.SequentialLR(
    optimizer, schedulers=[warmup, cosine], milestones=[warmup_steps]
)

for epoch in range(epochs):
    model.train()
    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)
        optimizer.zero_grad()
        z = model(images)
        loss = loss_fn(z, labels)
        loss.backward()
        optimizer.step()
        scheduler.step()
    print(f"epoch {epoch}: loss {loss.item():.4f}, lr {scheduler.get_last_lr()[0]:.6f}")

model.eval()

correct = 0
with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)
        pred = torch.argmax(model(images), dim=1)
        correct += (pred == labels).sum().item()
print(f"test accuracy: {correct / len(test_data):.4f}")

test_images, test_labels = next(iter(test_loader))
with torch.no_grad():
    z = model(test_images.to(device))
    pred = torch.argmax(z, dim=1).cpu()

plt.figure(figsize=(12, 12))
for i in range(64):
    plt.subplot(8, 8, i + 1)
    plt.imshow(test_images[i].squeeze(), cmap="gray")
    if test_labels[i] == pred[i]:
        color = "green"
    else:
        color = "red"
    plt.title(f"{test_labels[i].item()} → {pred[i].item()}", color=color)
    plt.axis("off")
plt.tight_layout()
plt.show()
```
