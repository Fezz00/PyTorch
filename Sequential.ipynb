```python
import numpy as np
import torch
import torch.nn as nn
```

```python
if(torch.cuda.is_available()):
    print(torch.cuda.get_device_name())
```

```python
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

model = nn.Linear(1,1)
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.05)
print(f"initial weight: {model.weight}\n")
for i in range(3000):
    model.zero_grad()

    z = model(x)
    loss = loss_fn(z, y)
    loss.backward()
    optimizer.step()
    if (i % 100) == 0:
        print(f"Loss {i}: {loss}")
print(f"\nw: {model.weight}\n\nb: {model.bias}")
```

```python
x = torch.tensor([[-2.0], [-1.0], [0.0], [1.0], [2.0]])
y = torch.tensor([[4.0], [1.0], [0.0], [1.0], [4.0]])

H = 10

model = nn.Sequential(
    nn.Linear(1, H),
    nn.ReLU(),
    nn.Linear(H, 1),
)

loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.05)

for i in range(3000):
    model.zero_grad()

    z = model(x)
    loss = loss_fn(z, y)
    loss.backward()
    optimizer.step()
    if (i % 100) == 0:
        print(f"Loss {i}: {loss}")
print(model(x))
```

```python
print(model(torch.tensor([[1.5]])))
print(model(torch.tensor([[3.0]])))
```
