```python
import numpy as np
import torch
```

```python
if(torch.cuda.is_available()):
    print(torch.cuda.get_device_name())
```

```python
w = torch.tensor(2.0, requires_grad=True)
x = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

z = w*x + b
print(z)
print(z.dtype)
z.backward()
print(w.grad, x.grad, b.grad)
```

```python
w = torch.tensor(2.0, requires_grad=True)
x = torch.tensor(3.0)
b = torch.tensor(-10.0, requires_grad=True)

z = w*x + b
a = torch.nn.functional.leaky_relu(z)
a.backward()
print(w.grad, b.grad)
```

```python
w = torch.tensor(2.0, requires_grad=True)
x = torch.tensor(3.0)
b = torch.tensor(-10.0, requires_grad=True)

z = w*x + b
a = torch.relu(z)
a.backward()
print(w.grad, b.grad)
```
