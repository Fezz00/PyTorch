```python
x = torch.tensor(3.0)
w = torch.tensor(1.0, requires_grad=True)
y = torch.tensor(6.0)

z = w * x
loss = (z - y) ** 2
loss.backward()

lr = 0.01
print("first loss: ", loss)
print("first w.grad: ", w.grad)

with torch.no_grad():
    w += - lr * w.grad
w.grad.zero_()
print("upadated w: ", w)

z = w * x
loss = (z - y) ** 2
loss.backward()
print("second w.grad: ", w.grad)
print("second loss: ",loss)
```

```python
x = torch.tensor(3.0)
w = torch.tensor(1.0, requires_grad=True)
y = torch.tensor(6.0)

lr = 0.01

for i in range(0,100):
    z = w * x
    loss = (z - y) ** 2
    loss.backward()
    if((i % 10) == 0):
        print(f"{i} loss: {loss}")

    with torch.no_grad():
        w += - lr * w.grad
    w.grad.zero_()
print(w)
```

```python
x = torch.tensor([1.0, 2.0, 3.0, 4.0])
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
y = torch.tensor([2.0, 4.0, 6.0, 8.0])

lr = 0.05

for i in range(3000):
    z = w * x + b
    loss = (z - y) ** 2
    loss = loss.mean()
    loss.backward()
    if((i % 100) == 0):
        print(f"{i} loss: {loss}")

    with torch.no_grad():
        w += - lr * w.grad
    w.grad.zero_()
    
    with torch.no_grad():
        b += - lr * b.grad
    b.grad.zero_()
print(w)
print(b)
```

```python
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

model = nn.Linear(1,1)
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.05) # SGD -> Adam

for i in range(3000):
    optimizer.zero_grad()
    z = model(x)
    loss = loss_fn(z, y)
    loss.backward()
    optimizer.step()
    if i % 100 == 0:
        print(f"{i} loss: {loss.item()}")

print(model.weight, model.bias)
```
